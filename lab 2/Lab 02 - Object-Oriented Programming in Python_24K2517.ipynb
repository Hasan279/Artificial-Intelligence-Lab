### Lab Task 1: AI-Powered Security Agents

```python
class SecurityAgent:
    def __init__(self, agent_id, name, status):
        self.agent_id = agent_id
        self.name = name
        self.status = status

class FirewallAgent(SecurityAgent):
    def monitor_traffic(self):
        print(f"[{self.name}] Monitoring network traffic (Status: {self.status})")

class MalwareDetectionAgent(SecurityAgent):
    def scan_files(self):
        print(f"[{self.name}] Scanning files for malware (Status: {self.status})")

class AutomationAgent(SecurityAgent):
    def run_automation(self):
        print(f"[{self.name}] Running AI automation routines (Status: {self.status})")

fw = FirewallAgent(1, "Network Firewall", "Active")
malware = MalwareDetectionAgent(2, "Malware Scanner", "Active")
auto = AutomationAgent(3, "System Optimizer", "Active")

fw.monitor_traffic()
malware.scan_files()
auto.run_automation()
```

### Lab Task 2: Threat Intelligence System

```python
class Threat:
    def __init__(self, threat_id, name, severity):
        self.threat_id = threat_id
        self.name = name
        self.severity = severity

class PhishingThreat(Threat):
    def analyze_email(self):
        print(f"[{self.name} | Severity: {self.severity}] Analyzing suspicious emails.")

class RansomwareThreat(Threat):
    def scan_files(self):
        print(f"[{self.name} | Severity: {self.severity}] Scanning file system for encrypted files.")

class BotnetThreat(Threat):
    def detect_traffic(self):
        print(f"[{self.name} | Severity: {self.severity}] Detecting unauthorized botnet network traffic.")

phish = PhishingThreat(101, "Phishing Attack", "High")
ransom = RansomwareThreat(102, "Ransomware Threat", "Critical")
bot = BotnetThreat(103, "Botnet Network", "Medium")

phish.analyze_email()
ransom.scan_files()
bot.detect_traffic()
```

### Lab Task 3: Automated Threat Response

```python
class ResponseAgent:
    def execute_response(self):
        print("Generic threat response.")

class AlertAgent(ResponseAgent):
    def execute_response(self):
        print("AlertAgent: Sending threat notification alerts to admin.")

class BlockAgent(ResponseAgent):
    def execute_response(self):
        print("BlockAgent: Blocking malicious IP addresses and connections.")

class RecoverAgent(ResponseAgent):
    def execute_response(self):
        print("RecoverAgent: Restoring affected systems and services.")

agents = [AlertAgent(), BlockAgent(), RecoverAgent()]

for agent in agents:
    agent.execute_response()
```

### Lab Task 5: Student Grading System

```python
class Student:
    def __init__(self, name, grade):
        self.name = name
        self.__grade = grade

    def set_grade(self, grade):
        self.__grade = grade

    def get_grade(self):
        return self.__grade

    def display_info(self):
        print(f"Student Name: {self.name}, Grade: {self.__grade}")

student = Student("Ali", 85)
student.display_info()

student.set_grade(92)
print("Grade accessed via getter:", student.get_grade())
student.display_info()

student.__grade = 50
print("Grade via getter after direct modification attempt:", student.get_grade())
print("Direct attribute student.__grade:", student.__grade)
```

### Task 1

```python
class Device:
    total_devices = 0

    def __init__(self, device_name, room, status="OFF"):
        self.device_name = device_name
        self.room = room
        self.status = status
        Device.total_devices += 1

    def turn_on(self):
        if self.status == "ON":
            print(f"{self.device_name} is already ON.")
        else:
            self.status = "ON"
            print(f"{self.device_name} turned ON.")

    def turn_off(self):
        if self.status == "OFF":
            print(f"{self.device_name} is already OFF.")
        else:
            self.status = "OFF"
            print(f"{self.device_name} turned OFF.")

    def display_status(self):
        print(f"{self.device_name} ({self.room}): {self.status}")


d1 = Device("Projector", "Room 101", "OFF")
d2 = Device("AC", "Room 101", "OFF")
d3 = Device("Smart Board", "Room 102", "ON")
d4 = Device("Lights", "Room 102", "OFF")

devices = [d1, d2, d3, d4]

print("Status before switching:")
for d in devices:
    d.display_status()

print("\nSwitching devices:")
for d in devices:
    d.turn_on()

print("\nStatus after switching:")
for d in devices:
    d.display_status()

# Both access the same class attribute because Python looks up the class namespace if not found in the instance.
print("\nClass access:", Device.total_devices)
print("Instance access:", d1.total_devices)
print(f"Total devices in the system: {Device.total_devices}")
```

### Task 2

```python
class LibraryItem:
    def __init__(self,title,item_id,status="Available"):
        self.title = title
        self.item_id = item_id
        self.status = status

    def issue_item(self):
        if self.status == "Available":
            self.status = "Not Available"
            print(f"Book {self.item_id} : {self.title} issued successfully")
        else:
            print(f"Book {self.item_id} : {self.title} is not available")

    def display_info(self):
        print(f"ID: {self.item_id} | Title: {self.title} | Status: {self.status}")

class Book(LibraryItem):
    def __init__(self, title, item_id, author, pages, status="Available"):
        super().__init__(title, item_id, status)
        self.author = author
        self.pages = pages

    def display_info(self):
        super().display_info()
        print(f"Author: {self.author} | Pages: {self.pages}")

class Ebook(LibraryItem):
        def __init__(self, title, item_id, file_size_mb, download_link, status="Available"):
            super().__init__(title,item_id,status)
            self.file_size_mb = file_size_mb
            self.download_link = download_link

        def display_info(self):
            super().display_info()
            print(f"File Size: {self.file_size_mb}MB | Link: {self.download_link}")

class Journal(LibraryItem):
    def __init__(self, title, item_id, volume, issue_date, status="Available"):
        super().__init__(title, item_id, status)
        self.volume = volume
        self.issue_date = issue_date

    def display_info(self):
        super().display_info()
        print(f"Volume: {self.volume} | Issue Date: {self.issue_date}")


class ReferenceBook(Book):
    def __init__(self, title, item_id, author, pages, status="Reference Only"):
        super().__init__(title, item_id, author, pages, status)

    def issue_item(self):
        print(f"Cannot issue '{self.title}'. Reference books are for library use only.")

# Create one object of each class
book = Book("Python Crash Course", "B101", "Eric Matthes", 544)
ebook = Ebook("Fluent Python", "E202", 15.5, "https://lib.fast.edu/dl/ebook202")
journal = Journal("IEEE AI Transactions", "J303", 42, "August 2026")
ref_book = ReferenceBook("Oxford English Dictionary", "R404", "Oxford Press", 21728)

items = [book, ebook, journal, ref_book]

print("=== Library Testing ===")
for item in items:
    print("-" * 40)
    item.display_info()
    item.issue_item()
```

```python
import math

class Shape:
    def area(self):
        print("Shape area is undefined")
        return 0

    def perimeter(self):
        print("Shape perimeter is undefined")
        return 0

    def describe(self, unit="cm", *extras):
        shape_name = self.__class__.__name__
        print(f"Shape: {shape_name} | Area: {self.area()} {unit}² | Perimeter: {self.perimeter()} {unit}")
        if extras:
            print(f"  Notes: {', '.join(extras)}")


class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height
    
    def area(self):
        return round(self.width * self.height, 2)
    
    def perimeter(self):
        return round(2 * (self.width + self.height), 2)


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return round(math.pi * (self.radius ** 2), 2)

    def perimeter(self):
        return round(2 * math.pi * self.radius, 2)


class Triangle(Shape):
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c

    def area(self):
        s = (self.a + self.b + self.c) / 2
        return round(math.sqrt(s * (s - self.a) * (s - self.b) * (s - self.c)), 2)

    def perimeter(self):
        return round(self.a + self.b + self.c, 2)


# Square demonstrates Duck Typing: it is NOT derived from Shape, but provides area() and 
# perimeter(), allowing it to be treated polymorphically in the loop without inheritance.
class Square:
    def __init__(self, side):
        self.side = side

    def area(self):
        return round(self.side ** 2, 2)

    def perimeter(self):
        return round(4 * self.side, 2)


# --- Main ---
rect = Rectangle(10, 5)
circle = Circle(7)
triangle = Triangle(3, 4, 5)
square = Square(6)

shapes = [rect, circle, triangle, square]

for s in shapes:
    name = s.__class__.__name__
    print(f"{name} -> Area: {s.area()}, Perimeter: {s.perimeter()}")

print()
rect.describe()
circle.describe("m", "Filled: Red", "Border: Solid")
triangle.describe("inches", "Right-angled triangle")
```

```python
class Account:
    def __init__(self, account_holder, initial_balance, pin):
        self.account_holder = account_holder
        self.__balance = initial_balance
        self.__pin = pin
        self._transaction_history = []
        self._transaction_history.append(f"Account opened with balance: ${initial_balance}")

    def check_pin(self, pin):
        return self.__pin == pin

    def get_balance(self, pin):
        if not self.check_pin(pin):
            print("Access Denied: Incorrect PIN.")
            return None
        return self.__balance

    def deposit(self, amount, pin):
        if not self.check_pin(pin):
            print("Deposit Failed: Incorrect PIN.")
            return False
        if amount <= 0:
            print(f"Deposit Failed: Amount must be positive.")
            return False
        
        self.__balance += amount
        self._transaction_history.append(f"Deposited: ${amount} | Balance: ${self.__balance}")
        print(f"Deposited ${amount}. Balance: ${self.__balance}")
        return True

    def withdraw(self, amount, pin):
        if not self.check_pin(pin):
            print("Withdrawal Failed: Incorrect PIN.")
            return False
        if amount <= 0:
            print(f"Withdrawal Failed: Amount must be positive.")
            return False
        if amount > self.__balance:
            print(f"Withdrawal Failed: Insufficient funds.")
            return False

        self.__balance -= amount
        self._transaction_history.append(f"Withdrew: ${amount} | Balance: ${self.__balance}")
        print(f"Withdrew ${amount}. Balance: ${self.__balance}")
        return True

    def print_statement(self, pin):
        if not self.check_pin(pin):
            print("Access Denied: Incorrect PIN.")
            return
        
        print(f"Statement for {self.account_holder}:")
        for record in self._transaction_history:
            print(f" - {record}")


acc = Account("Hasan", 1000, 1234)

acc.deposit(500, 1234)
acc.deposit(-100, 1234)
acc.withdraw(300, 1234)
acc.withdraw(2000, 1234)
acc.withdraw(200, 9999)
print("Balance:", acc.get_balance(1234))

acc.print_statement(1234)

try:
    print(acc.__balance)
except AttributeError as e:
    print("Direct access failed:", e)

print("Mangled name access:", acc._Account__balance)

# Python renames private variables using name mangling (_Account__balance) rather than enforcing strict access control. Because the variable is still reachable through its mangled name, privacy in Python is a convention rather than true security.
```
