#### Tuple — the immutable sibling of the list

Everything a list can *read*, a tuple can read too. What a tuple cannot do is change. The last cell of this section deliberately triggers a `TypeError` and catches it, so that you can see the error message without stopping the notebook.

```python
my_tuple  = ('abcd', 786, 2.23, 'khadeejah', 70.2)
tinytuple = (123, 'khadeejah')

print(my_tuple)                # Prints the complete tuple
print(my_tuple[0])             # Prints the first element of the tuple
print(my_tuple[1:3])           # Prints the elements from the 2nd till the 3rd
print(my_tuple[2:])            # Prints the elements starting from the 3rd element
print(tinytuple * 2)           # Prints the tuple two times
print(my_tuple + tinytuple)    # Prints the concatenated tuple
```

```python
# A list can be updated, but a tuple cannot.
my_list = ['abcd', 786, 2.23, 'khadeejah', 70.2]
my_list[2] = 1000              # Valid syntax with a list
print('Updated list :', my_list)

# The line below is invalid for a tuple. We catch the error on purpose so the
# notebook keeps running and you can read the exact message Python produces.
try:
    my_tuple[2] = 1000
except TypeError as error:
    print('TypeError :', error)

# The correct way to "change" a tuple is to build a new one
my_tuple = my_tuple[:2] + (1000,) + my_tuple[3:]
print('Rebuilt tuple :', my_tuple)
```

### 4.6 Functions

A function is a block of reusable code that performs one related action. It is defined with `def` and runs only when it is **called**.

```python
# Function definition
def non_para_func():
    print("This is a non-parametric function")

non_para_func()          # Function call
```

```python
# Function definition with parameters
def para_func(a, b):
    """This function takes the values of a and b provided by the caller."""
    return a + b

print('para_func(1, 2) returns :', para_func(1, 2))
```

#### The `return` statement

A function that has no `return` statement (or a bare `return`) gives back `None`. You cannot store anything useful from such a function.

```python
# Function definition
def add_numbers(arg1, arg2):
    # Add both the parameters and return the result
    total = arg1 + arg2
    print("Inside the function : ", total)
    return total

# Now the function can be called
total = add_numbers(10, 20)
print("Outside the function : ", total)

# A function that returns nothing
def print_message(message):
    print('Print Statement : ', message)
    return                    # This is equivalent to 'return None'

result = print_message("Hello, World!")
print(result)                 # Output: None
```

#### 4.6.1 Types of functions (based on arguments)

* **Positional** — values are matched strictly by position.
* **Default** — a parameter has a fallback value used when the caller omits it.
* **Keyword** — the caller names the parameter, so order stops mattering.

```python
# Positional arguments - values are mapped strictly by position
def details(name, age):
    print('Name : ', name)
    print('Age  : ', age)

details(23, 'Khadeejah')     # name = 23 and age = 'Khadeejah' - the order is wrong!
print('---')
details('Khadeejah', 23)     # this is the correct call
```

```python
# Default argument
def details(name, age=23):
    print('Name : ', name)
    print('Age  : ', age)

details('Khadeejah')         # age takes its default value 23
print('---')
details('Khadeejah', 30)     # the default is overridden
```

```python
# Keyword arguments - order no longer matters
def details(name, age, height):
    print('Name   : ', name)
    print('Age    : ', age)
    print('Height : ', height)

details(name='Khadeejah', age=23, height=5.5)
print('---')
details(age=23, height=5.5, name='Khadeejah')   # same result, different order
```

#### 4.6.2 Types of functions (based on number of parameters)

* **Fixed length** — the number of arguments is decided when the function is written.
* **Variable length** — `*args` collects extra positional arguments into a tuple, `**kwargs` collects extra keyword arguments into a dictionary.

```python
# Function with fixed parameters
def add(a, b):
    '''a and b are fixed parameters'''
    return a + b

print('add(1, 2) =', add(1, 2))

# Providing extra arguments raises a TypeError - caught here so the notebook continues
try:
    add(1, 2, 3)
except TypeError as error:
    print('TypeError :', error)
```

```python
# Variable length arguments (*args) - collected into a tuple
def details(*numbers):
    print('Numbers : ', numbers, type(numbers))

details(1, 2, 3)
details(1, 2, 3, 4, 5)       # any number of arguments is accepted

# Keyword variable length arguments (**kwargs) - collected into a dictionary
def details(**name):
    print('Names : ', name, type(name))

details(n1='Khadeejah', n2='Maryam', n3='Hira')
```

## 5. Putting it together: a rule-based AI agent

The earliest AI systems were **rule-based agents**: they map an input pattern to an action using a fixed knowledge base. Everything below uses only what this lab has covered — a dictionary, a loop, a conditional and functions with default, variable length and keyword parameters.

This is a miniature version of Lab Task 5, provided so that you can see the shape of the solution before you write your own.

```python
# The knowledge base: each keyword maps to the answer the agent should give
knowledge_base = {
    'name':  'My name is Hasan Waseem',
    'age': 'I am 20 years old',
    'department': 'AI&DS',
    'project': 'The semester project is demonstrated in Week 15.',
}


def match_rule(question, rules, fallback='Sorry, that is outside my knowledge base.'):
    """Return the answer whose keyword appears in the question, else the fallback.

    'fallback' is a default argument, so the caller may leave it out.
    """
    question = question.lower()
    for keyword in rules:
        if keyword in question:
            return rules[keyword]
    return fallback


def session_report(*questions, **stats):
    """*questions collects the asked questions, **stats collects named counters."""
    print('--- Session report ---')
    for key in stats:
        print(key, ':', stats[key])
    print('Questions asked :')
    for q in questions:
        print('   -', q)
```

```python
# A scripted session so the notebook runs without waiting for input.
# In the lab, replace this list with a while loop that calls input().
user_name = 'Khadeejah'
asked = ['What is my name and age?',
         'What is my department?',
         'Tell me about the project',
         'What is the weather today?']

print('Hello', user_name + '! Ask me about the AI Lab.')

matched = 0
unmatched = 0

for question in asked:
    answer = match_rule(question, knowledge_base)
    print()
    print('You    :', question)
    print('Agent  :', answer)
    if answer in knowledge_base.values():
        matched = matched + 1
    else:
        unmatched = unmatched + 1

print()
session_report(*asked, analyst=user_name, age=23, matched=matched, unmatched=unmatched)
```

### Lab Task 3: Text Analyser for an AI Pre-processing Step

```python
# Lab Task 3 - your solution here
sentence = input("Enter a sentence: ")

vowels_count = 0
consonants_count = 0
vowels = "aeiouAEIOU"

for ch in sentence:
    if ch.isalpha():
        if ch in vowels:
            vowels_count += 1
        else:
            consonants_count += 1

unique_chars = set()
for ch in sentence:
    if ch != " ":
        unique_chars.add(ch)

char_freq = {}
for ch in sentence:
    if ch in char_freq:
        char_freq[ch] += 1
    else:
        char_freq[ch] = 1

repeated_chars = {}
for ch in char_freq:
    if char_freq[ch] > 1:
        repeated_chars[ch] = char_freq[ch]

reversed_sentence = ""
i = len(sentence) - 1
while i >= 0:
    reversed_sentence += sentence[i]
    i -= 1

first_five = sentence[:5]
last_five = sentence[-5:]

print("Original Sentence:", sentence)
print("Vowels Count:", vowels_count)
print("Consonants Count:", consonants_count)
print("Unique Characters:", unique_chars)
print("Number of Unique Characters:", len(unique_chars))
print("Repeated Characters:", repeated_chars)
print("Reversed Sentence:", reversed_sentence)
print("First 5 Characters:", first_five)
print("Last 5 Characters:", last_five)
```

### Lab Task 4: Number Toolkit Using Functions

```python
# Lab Task 4 - your solution here
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True

def digit_sum(n):
    n = abs(n)
    total = 0
    while n > 0:
        total += n % 10
        n = n // 10
    return total

def classify(n, threshold=100):
    if n <= threshold:
        return "small"
    else:
        return "large"

def summarise(*numbers):
    if len(numbers) == 0:
        return None
    smallest = numbers[0]
    largest = numbers[0]
    total = 0
    count = 0
    for num in numbers:
        if num < smallest:
            smallest = num
        if num > largest:
            largest = num
        total += num
        count += 1
    average = total / count
    return {
        "smallest": smallest,
        "largest": largest,
        "average": average
    }

def describe(**details):
    for key, value in details.items():
        print(key, ":", value)
    return None

primes = []
for x in range(1, 30):
    if is_prime(x):
        primes.append(x)
print("Primes from 1 to 30:", primes)

print("Digit sum of 9876:", digit_sum(9876))

print("Classify with positional argument:", classify(150, 100))
print("Classify with default argument:", classify(50))
print("Classify with keyword argument:", classify(n=120, threshold=200))

summary_result = summarise(15, 3, 42, 8, 27, 91, 14)
print("Summary Statistics:", summary_result)

print("Student Details:")
describe(name="Khadeejah", age=23)
```

### Lab Task 5: Rule-Based AI Advisor (Capstone)

```python
# Lab Task 5 - your solution here
def match_rule(question, knowledge_base):
    lower_question = question.lower()
    for keyword in knowledge_base:
        if keyword in lower_question:
            return knowledge_base[keyword]
    return None

def session_report(*questions, **stats):
    print("\n--- Session Report ---")
    print("Questions Asked:")
    for q in questions:
        print("-", q)
    for key, value in stats.items():
        print(key, ":", value)
    return None

knowledge_base = {
    "python": "Python is the primary language used for AI in this lab.",
    "grading": "Grading consists of in-lab tasks, home tasks, and lab assessments.",
    "project": "The lab project involves building a practical AI solution or agent.",
    "timing": "Lab sessions are conducted weekly for three hours.",
    "attendance": "A minimum of 80% attendance is required to pass the lab."
}

user_name = input("Enter your name: ")
user_age = input("Enter your age: ")
print("Welcome", user_name + "!", "AI Advisor is ready. Type 'exit' to end the session.")

questions_list = []
matched_count = 0
unmatched_count = 0
fallback_answer = "I do not have information on this topic yet. Please contact the instructor."

while True:
    user_question = input("\nAsk a question: (or type exit to quit)")
    if user_question.strip().lower() == "exit":
        break

    questions_list.append(user_question)
    answer = match_rule(user_question, knowledge_base)

    if answer is not None:
        print("Advisor:", answer)
        matched_count += 1
    else:
        print("Advisor:", fallback_answer)
        unmatched_count += 1

session_report(*questions_list, Matched_Rules=matched_count, Unmatched_Rules=unmatched_count)
```
